#### Task 03  ·  Multi-Source ETL

```python
import pandas as pd
import requests
import mysql.connector

# EXTRACT: Fetch users and posts from API endpoints
users_url = "https://jsonplaceholder.typicode.com/users"
posts_url = "https://jsonplaceholder.typicode.com/posts"

users_response = requests.get(users_url)
posts_response = requests.get(posts_url)

users_data = users_response.json()
posts_data = posts_response.json()
```

```python
# Create df_users with normalized address.city
df_users = pd.json_normalize(users_data)
df_users = df_users[['id', 'name', 'email', 'address.city']]
df_users.columns = ['id', 'name', 'email', 'city']

print("Users DataFrame:")
print(df_users.head())
```

```python
# Create df_posts with renamed userId to id
df_posts = pd.DataFrame(posts_data)
df_posts = df_posts[['userId', 'title']]
df_posts = df_posts.rename(columns={'userId': 'id'})

print("Posts DataFrame:")
print(df_posts.head())
```

```python
# TRANSFORM: Merge on 'id'
df_merged = pd.merge(df_users, df_posts, on='id', how='left')

# Count posts per user
post_count = df_posts.groupby('id').size().reset_index(name='post_count')
df_users = pd.merge(df_users, post_count, on='id', how='left')

print("Merged DataFrame with post_count:")
print(df_users.head())
```

```python
# CLEAN: lowercase email, strip whitespace from name and city, drop nulls
df_users['email'] = df_users['email'].str.lower()
df_users['name'] = df_users['name'].str.strip()
df_users['city'] = df_users['city'].str.strip()

df_users = df_users.dropna()

print("Cleaned DataFrame:")
print(df_users.head())
```

```python
# LOAD: Save to CSV
df_users.to_csv('merged_data.csv', index=False)
print("Saved to merged_data.csv")
```

```python
# Save to MySQL database
conn = mysql.connector.connect(
    host='localhost',
    user='root',
    password='root',
)

cursor = conn.cursor()
cursor.execute('CREATE DATABASE IF NOT EXISTS etl_db')
cursor.execute('USE etl_db')
# Create table
cursor.execute('''
    CREATE TABLE IF NOT EXISTS users (
        id INT PRIMARY KEY,
        name VARCHAR(255),
        email VARCHAR(255),
        city VARCHAR(255),
        post_count INT
    )
''')

# Clear existing data
cursor.execute('DELETE FROM users')

# Insert data
for index, row in df_users.iterrows():
    cursor.execute('''
        INSERT INTO users (id, name, email, city, post_count) 
        VALUES (%s, %s, %s, %s, %s)
    ''', (row['id'], row['name'], row['email'], row['city'], row['post_count']))

conn.commit()
conn.close()
print("Saved to MySQL database (etl_db.users)")
```

```python
# Top 3 most active users
top_3_users = df_users.nlargest(3, 'post_count')[['id', 'name', 'email', 'post_count']]
print("\nTop 3 Most Active Users:")
print(top_3_users)
```
