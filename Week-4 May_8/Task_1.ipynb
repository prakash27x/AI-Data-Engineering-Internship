#### Task 01  ·  Clean a Messy CSV

```python
import pandas as pd
df = pd.read_csv("messy_students_data.csv")
```

```python
df['Name'] = df['Name'].str.title().str.strip().str.strip(" '\" ")
df['Name']
```

```python
df['Gender'] = ( df['Gender'].astype(str).str.strip().str.title()
    .replace({
        'M': 'Male',
        'F': 'Female',
        '0': None
    })
)
```

```python
df["Gender"]
```

```python
df["Grade"] = (
    df["Grade"]
    .astype(str)
    .str.replace("Grade", "", regex=False)
    .str.strip()
    .str.lstrip("0")
)
```

```python
df["Grade"] = df["Grade"].astype(int)
df['Grade']
```

```python
df['Math'] = df['Math'].astype(str).str.replace("marks", "", regex=False).str.strip()
df['Math'] = df['Math'].astype(float)
df['Math']
```

```python
df['Science'] = df['Science'].astype(str).str.replace("marks", "", regex=False).str.strip()
df['Science'] = df['Science'].astype(float)
df['Science']
```

```python
df['English'] = df['English'].astype(str).str.replace("marks", "", regex=False).str.strip()
df['English'] = df['English'].astype(float)
df['English']
```

```python
df.info
```

```python
df.isnull()
```

```python
df = df.dropna()
```

```python
df.isnull()
```

```python
# Calculate average score and assign grades
df['score'] = (df['Math'] + df['Science'] + df['English']) / 3

def assign_grade(score):
    if score >= 90:
        return 'A'
    elif score >= 75:
        return 'B'
    elif score >= 50:
        return 'C'
    else:
        return 'F'

df['grade'] = df['score'].apply(assign_grade)
df[['Name', 'Math', 'Science', 'English', 'score', 'grade']]
```

### export to csv file

```python
df.to_csv('clean_students_data.csv', index=False)
```
