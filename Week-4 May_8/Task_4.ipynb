#### Task 04  ·  Transform & Enrich

```python
import pandas as pd

# Load clean_students.csv from Task 01
df = pd.read_csv('clean_students_data.csv')
df.head()
```

```python
# Add grade column: A (≥90), B (≥75), C (≥60), D (≥50), F (<50)
def assign_grade(score):
    if score >= 90:
        return 'A'
    elif score >= 75:
        return 'B'
    elif score >= 60:
        return 'C'
    elif score >= 50:
        return 'D'
    else:
        return 'F'

df['grade'] = df['score'].apply(assign_grade)

print("After adding grade column:")
print(df[['score', 'grade']].head())
```

```python
# Add passed column: True if score >= 50, False otherwise
df['passed'] = df['score'] >= 50

print("After adding passed column:")
print(df[['score', 'grade', 'passed']].head())
```

```python
# Add score_category column: 'High' (≥80), 'Medium' (50–79), 'Low' (<50)
def categorize_score(score):
    if score >= 80:
        return 'High'
    elif score >= 50:
        return 'Medium'
    else:
        return 'Low'

df['score_category'] = df['score'].apply(categorize_score)

print("After adding score_category column:")
print(df[['score', 'score_category']].head())
```

```python
# Add rank column: rank all students by score — highest score gets rank 1
df['rank'] = df['score'].rank(ascending=False).astype(int)

print("After adding rank column:")
print(df[['score', 'rank']].head())
```

```python
# Group by grade — print count, mean score, and min/max score per grade
grade_summary = df.groupby('grade')['score'].agg(['count', 'mean', 'min', 'max'])

print("\nGroupBy Grade Summary:")
print(grade_summary)
```

```python
# Sort by rank and reset index
df = df.sort_values('rank').reset_index(drop=True)

print(df[['score', 'grade', 'rank']].head(10))
```

```python
# Save enriched data to enriched_students.csv
df.to_csv('enriched_students.csv', index=False)
```
