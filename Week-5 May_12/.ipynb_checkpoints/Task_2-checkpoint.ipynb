#### Task 2: Data Quality Audit on Top of an ETL Pipeline

This notebook fetches 100 posts from JSONPlaceholder, audits the raw data before any cleaning, applies the required transformations, generates a structured audit report, and loads the clean dataset into MySQL.

```python
from pathlib import Path

import mysql.connector
import pandas as pd
import requests
from IPython.display import display

API_URL = "https://jsonplaceholder.typicode.com/posts"
RAW_LIMIT = 100
AUDIT_REPORT_PATH = Path.cwd() / "task_2_audit_report.csv"
EXPECTED_SCHEMA = {"userId": "int64", "id": "int64", "title": "string", "body": "string"}

def fetch_posts(limit: int = RAW_LIMIT) -> pd.DataFrame:
    response = requests.get(API_URL, params={"_limit": limit}, timeout=20)
    response.raise_for_status()
    return pd.DataFrame(response.json()[:limit])[["userId", "id", "title", "body"]]
```

### Audit

```python
def build_pre_clean_audit(df: pd.DataFrame) -> dict:
    null_counts = df.isna().sum().rename_axis("column").reset_index(name="null_count")
    duplicate_row_count = int(df.duplicated().sum())

    type_checks = pd.DataFrame(
        [
            {
                "column": "userId",
                "expected_type": EXPECTED_SCHEMA["userId"],
                "actual_type": str(df["userId"].dtype),
                "mismatch": not pd.api.types.is_integer_dtype(df["userId"]),
            },
            {
                "column": "id",
                "expected_type": EXPECTED_SCHEMA["id"],
                "actual_type": str(df["id"].dtype),
                "mismatch": not pd.api.types.is_integer_dtype(df["id"]),
            },
            {
                "column": "title",
                "expected_type": EXPECTED_SCHEMA["title"],
                "actual_type": str(df["title"].dtype),
                "mismatch": not pd.api.types.is_string_dtype(df["title"]),
            },
            {
                "column": "body",
                "expected_type": EXPECTED_SCHEMA["body"],
                "actual_type": str(df["body"].dtype),
                "mismatch": not pd.api.types.is_string_dtype(df["body"]),
            },
        ]
    )

    out_of_range_checks = pd.DataFrame(
        [
            {"column": "userId", "rule": "between 1 and 10", "issue_count": int((~df["userId"].between(1, 10)).sum())},
            {"column": "id", "rule": "between 1 and 100", "issue_count": int((~df["id"].between(1, 100)).sum())},
        ]
    )

    cleaned_titles = df["title"].fillna("").astype(str).str.strip()
    title_format_mask = cleaned_titles != cleaned_titles.str.title()

    issue_summary = pd.DataFrame(
        [
            {"issue_type": "Null values", "detected_count": int(df.isna().sum().sum())},
            {"issue_type": "Duplicate rows", "detected_count": duplicate_row_count},
            {"issue_type": "Type mismatches", "detected_count": int(type_checks["mismatch"].sum())},
            {"issue_type": "Out-of-range values", "detected_count": int(out_of_range_checks["issue_count"].sum())},
            {"issue_type": "Inconsistent string formats", "detected_count": int(title_format_mask.sum())},
        ]
    )

    return {
        "null_counts": null_counts,
        "type_checks": type_checks,
        "out_of_range_checks": out_of_range_checks,
        "inconsistent_string_examples": df.loc[title_format_mask, ["id", "title"]].head(10),
        "issue_summary": issue_summary,
        "total_issues": int(issue_summary["detected_count"].sum()),
    }
```

### Clean

```python
def clean_posts(df: pd.DataFrame) -> tuple[pd.DataFrame, dict]:
    cleaned = df.copy()
    cleaned["title"] = cleaned["title"].fillna("").astype(str).str.strip().str.title()
    cleaned["body"] = cleaned["body"].fillna("").astype(str).str.strip()

    title_fixed_count = int((df["title"].fillna("").astype(str).str.strip() != cleaned["title"]).sum())
    cleaned = cleaned[(cleaned["title"] != "") & (cleaned["body"] != "")]

    before_duplicates = len(cleaned)
    cleaned = cleaned.drop_duplicates(subset=["id"])
    duplicates_removed = before_duplicates - len(cleaned)

    before_range = len(cleaned)
    cleaned = cleaned[cleaned["userId"].between(1, 10) & cleaned["id"].between(1, 100)]
    out_of_range_removed = before_range - len(cleaned)

    cleaned["word_count"] = cleaned["title"].str.split().str.len().fillna(0).astype(int) + cleaned["body"].str.split().str.len().fillna(0).astype(int)
    cleaned["title_rank"] = cleaned["word_count"].rank(method="dense", ascending=False).astype(int)
    cleaned = cleaned.sort_values(["title_rank", "id"]).reset_index(drop=True)

    fix_summary = {
        "title_fixed_count": title_fixed_count,
        "duplicates_removed": duplicates_removed,
        "out_of_range_removed": out_of_range_removed,
        "type_coercions": 0,
    }
    return cleaned, fix_summary


def build_audit_report(pre_clean_audit: dict, fix_summary: dict, before_rows: int, after_rows: int) -> pd.DataFrame:
    report = pre_clean_audit["issue_summary"].copy()
    report["fixed_count"] = [0, fix_summary["duplicates_removed"], fix_summary["type_coercions"], fix_summary["out_of_range_removed"], fix_summary["title_fixed_count"]]
    report["before_row_count"] = before_rows
    report["after_row_count"] = after_rows
    return report
```

### Load

```python
def load_to_mysql(df: pd.DataFrame) -> int:
    mysql_config = {
        "host": os.getenv("MYSQL_HOST", "localhost"),
        "port": int(os.getenv("MYSQL_PORT", "3306")),
        "user": os.getenv("MYSQL_USER", "root"),
        "password": os.getenv("MYSQL_PASSWORD", "MONOuba123!@#"),
        "database": os.getenv("MYSQL_DATABASE", "week5_audit"),
    }

    connection = mysql.connector.connect(
        host=mysql_config["host"],
        port=mysql_config["port"],
        user=mysql_config["user"],
        password=mysql_config["password"],
    )
    cursor = connection.cursor()
    cursor.execute(f"CREATE DATABASE IF NOT EXISTS `{mysql_config['database']}`")
    cursor.execute(f"USE `{mysql_config['database']}`")
    cursor.execute(
        """
        CREATE TABLE IF NOT EXISTS clean_posts (
            id INT PRIMARY KEY,
            userId INT NOT NULL,
            title VARCHAR(255) NOT NULL,
            body TEXT NOT NULL,
            word_count INT NOT NULL,
            title_rank INT NOT NULL
        )
        """
    )

    insert_sql = """
        INSERT INTO clean_posts (id, userId, title, body, word_count, title_rank)
        VALUES (%s, %s, %s, %s, %s, %s)
        ON DUPLICATE KEY UPDATE
            userId = VALUES(userId),
            title = VALUES(title),
            body = VALUES(body),
            word_count = VALUES(word_count),
            title_rank = VALUES(title_rank)
    """
    rows = list(df[["id", "userId", "title", "body", "word_count", "title_rank"]].itertuples(index=False, name=None))
    cursor.executemany(insert_sql, rows)
    connection.commit()
    cursor.close()
    connection.close()
    return len(rows)
```

```python
raw_posts = fetch_posts()
raw_posts.head()
print(f"Fetched {len(raw_posts)} rows and {len(raw_posts.columns)} columns from JSONPlaceholder.")
print(raw_posts.dtypes)
```

```python
pre_clean_audit = build_pre_clean_audit(raw_posts)

print("Null counts by column:")
display(pre_clean_audit["null_counts"])

print("Type checks:")
display(pre_clean_audit["type_checks"])

print("Out-of-range checks:")
display(pre_clean_audit["out_of_range_checks"])

print("Examples of inconsistent string formats:")
display(pre_clean_audit["inconsistent_string_examples"])

print("Issue summary before cleaning:")
display(pre_clean_audit["issue_summary"])
print(f"Total issues detected before cleaning: {pre_clean_audit['total_issues']}")
```

```python
clean_posts_df, fix_summary = clean_posts(raw_posts)
clean_posts_df.head()
print(f"Rows before cleaning: {len(raw_posts)}")
print(f"Rows after cleaning: {len(clean_posts_df)}")
print(fix_summary)
```

```python
audit_report = build_audit_report(pre_clean_audit, fix_summary, len(raw_posts), len(clean_posts_df))

audit_report.to_csv(AUDIT_REPORT_PATH, index=False)
print(f"Audit report saved to: {AUDIT_REPORT_PATH}")
print(f"Total issues detected: {pre_clean_audit['total_issues']}")
print(f"Total issues fixed: {int(audit_report['fixed_count'].sum())}")
print(f"Before row count: {len(raw_posts)}")
print(f"After row count: {len(clean_posts_df)}")
display(audit_report)
```

## MySQL Load

```python
try:
    rows_loaded = load_to_mysql(clean_posts_df)
    print(f"Rows written to MySQL: {rows_loaded}")
except mysql.connector.Error as exc:
    print("MySQL load could not be completed in this environment.")
    print("Set the MySQL environment variables and make sure the server is running, then rerun this cell.")
    print(exc)
```
