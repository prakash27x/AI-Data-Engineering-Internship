### EDA Summary
This dataset contains both numeric and categorical columns. During EDA, missing values, data distributions, category frequencies, and possible outliers were analyzed using summary statistics and visualizations.

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("messy_students_data.csv")
print("Dataset Loaded Successfully")
```

```python
# VIEW FIRST 5 ROW
df.head()
```

```python
# STEP 2 — SHAPE OF DATASET
print("Dataset Shape:(row, column)")
print(df.shape)
```

```python
df.info()

# Observation:
# Check:
# - null values
# - column names
# - data types
```

```python
print(df.dtypes)
```

```python
print(df.isnull().sum())
```

```python
#Descriptive statistics
print(df.describe())

# Observation:
# Check:
# - mean
# - min
# - max
# - standard deviation
```

```python
print(df['Gender'].value_counts())
```

```python
# SELECT NUMERIC COLUMNS

numeric_columns = df.select_dtypes(
    include=['int64', 'float64']
).columns

print(numeric_columns)
```

```python
# HISTOGRAM FOR EVERY NUMERIC COLUMN

for col in numeric_columns:

    plt.figure(figsize=(6, 4))

    df[col].hist(bins=20)

    plt.title(f"Histogram of {col}")

    plt.xlabel(col)

    plt.ylabel("Frequency")

    # Save chart
    plt.savefig(f"{col}_histogram.png")

    plt.show()

# Observation:
# Check if data is normal or skewed
```

```python
# BOXPLOT FOR FIRST 2 NUMERIC COLUMNS


for col in numeric_columns[:2]:

    plt.figure(figsize=(6, 4))

    plt.boxplot(df[col].dropna())

    plt.title(f"Boxplot of {col}")

    # Save chart
    plt.savefig(f"{col}_boxplot.png")

    plt.show()

# Observation:
# Dots outside box are outliers
```
